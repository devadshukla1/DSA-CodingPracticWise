```python
def CountingSort(arr):
    maxval = max(arr)
    count = [0] * (maxval + 1)  # Count array

    while len(arr) > 0:
        num = arr.pop(0)  # Pop out first element
        count[num] += 1

    for i in range(len(count)):
        while count[i] > 0:
            arr.append(i)
            count[i] -= 1

    return arr


mylist = [4, 2, 2, 6, 3, 3, 1, 6, 5, 2, 3]

mysortedlist = CountingSort(mylist)

print(mysortedlist)
```
